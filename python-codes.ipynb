```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine


,DATABASE_URL = "postgresql://postgres:2038@localhost:5432/rent_analysis"
engine = create_engine(DATABASE_URL)

query = "SELECT * FROM rent_analysis"
df = pd.read_sql(query, engine)

df.head()
```

```python
# Grafik stilini belirleyelim
sns.set_theme(style="whitegrid")

# Şehirlere göre ortalama kira fiyatlarını çizelim
plt.figure(figsize=(12, 6))
plot = sns.barplot(x='city', y='price_monthly', data=df, estimator='mean', palette='magma')

plt.title('Average Monthly Rent by City', fontsize=16)
plt.xlabel('City', fontsize=12)
plt.ylabel('Average Rent ($)', fontsize=12)
plt.xticks(rotation=45)

# Grafiği bilgisayarına kaydet (GitHub README'ye eklemek için)
plt.savefig('average_rent_by_city.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='size_sqm', y='price_monthly', hue='city', size='building_age', data=df, palette='viridis')

plt.title('Price vs Square Meters (Colored by City)', fontsize=15)
plt.show()
```
